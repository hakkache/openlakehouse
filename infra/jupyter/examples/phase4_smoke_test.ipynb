# OpenLakehouse — Phase 4 Notebook Smoke Test

Real PySpark execution against the live Spark cluster, reading the Iceberg
table created in Phase 3 via the Polaris REST catalog / MinIO storage.

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("openlakehouse-notebook-test").getOrCreate()
print("PHASE4_SPARK_SESSION_OK", spark.version)
```

```python
count = spark.range(50).count()
assert count == 50
print("PHASE4_RANGE_OK", count)
```

```python
df = spark.table("catalog.bronze.test")
row_count = df.count()
print("PHASE4_ICEBERG_READ_OK", row_count)
df.show()
```

```python
spark.sql("CREATE NAMESPACE IF NOT EXISTS catalog.bronze")
notebook_df = spark.range(5).withColumnRenamed("id", "id")
notebook_df.writeTo("catalog.bronze.from_notebook").createOrReplace()
written_count = spark.table("catalog.bronze.from_notebook").count()
assert written_count == 5
print("PHASE4_ICEBERG_WRITE_OK", written_count)
```
